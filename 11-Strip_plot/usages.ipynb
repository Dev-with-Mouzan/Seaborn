# Strip Plot — When to Use It

A strip plot places every observation along a categorical axis with jitter, showing the full distribution of points.

## Real-World Use Cases

**1. Scores by gender** — Show every student's score per gender to see the actual data density.

**2. Ratings by product category** — See each rating point for every category, including overlapping clusters.

**3. Counts per group** — Reveal where points cluster when many observations share similar values.
