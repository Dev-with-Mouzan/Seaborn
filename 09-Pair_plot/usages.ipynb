# Pair Plot — When to Use It

A pair plot shows pairwise scatter plots and distributions for every pair of numeric columns in one figure.

## Real-World Use Cases

**1. Exam score relationships** — Explore how math, reading and writing scores relate, colored by gender.

**2. Quick EDA of numeric data** — Scan all pairwise relationships of a small numeric dataset before modeling.

**3. Feature exploration** — Check which variables move together before building a model.
