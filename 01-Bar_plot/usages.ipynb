# Bar Plot — When to Use It

A bar plot shows the average of a numeric variable for each category, complete with error bars, making group comparisons easy.

## Real-World Use Cases

**1. Comparing average math scores by gender** — Compare the mean math score (with confidence intervals) for male vs female students.

**2. Sales by region** — Compare average or total sales per region to see which performs best.

**3. Team productivity** — Compare average output per team with error bars to show variability.
