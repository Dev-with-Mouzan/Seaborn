# Multiplot — When to Use It

A multiplot (FacetGrid) draws many small axes conditioned on subsets of the data, great for small-multiples comparisons.

## Real-World Use Cases

**1. Score distributions by gender** — Show a separate distribution subplot for each gender.

**2. Grid of scatter plots per category** — Compare relationships within each category side by side.

**3. Small multiples for groups** — Spot patterns across many groups without overcrowding one figure.
