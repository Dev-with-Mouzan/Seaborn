# Box Plot — When to Use It

A box plot summarizes the distribution of a numeric variable per category using quartiles, whiskers and outliers.

## Real-World Use Cases

**1. Reading scores by lunch type** — Compare the spread and median of reading scores across lunch categories and spot outliers.

**2. Salaries by department** — Compare pay ranges across departments to detect compensation differences.

**3. Sensor readings by machine state** — Compare measurement distributions across operating states and find unusual values.
