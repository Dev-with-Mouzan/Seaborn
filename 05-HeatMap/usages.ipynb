# Heat Map — When to Use It

A heat map colors a matrix by value, making patterns like correlations instantly visible.

## Real-World Use Cases

**1. Correlation matrix of exam scores** — See how math, reading and writing scores correlate with each other.

**2. Confusion matrix** — Visualize classification results so correct and misclassified cells stand out.

**3. Grid values** — Display data like activity by day of week and hour to reveal patterns.
