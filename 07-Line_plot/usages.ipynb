# Line Plot — When to Use It

A line plot shows how a variable changes over a continuous axis such as time or another ordered value.

## Real-World Use Cases

**1. Scores vs study hours** — Show the trend of exam scores as study time increases.

**2. Website traffic over time** — Track daily or monthly traffic to spot trends and seasonality.

**3. Temperature over months** — Visualize how temperature changes across the year.
