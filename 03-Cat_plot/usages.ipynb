# Cat Plot — When to Use It

sns.catplot is a flexible categorical plot that can render strip, swarm, box, violin and bar charts, ideal for comparing groups.

## Real-World Use Cases

**1. Writing scores by test prep** — Explore how writing scores vary between students who did and did not complete a test prep course.

**2. Scores across race/ethnicity** — Compare exam scores across race/ethnicity groups with several categorical plot styles.

**3. Faceting by gender** — Draw separate categorical plots for each gender to compare patterns side by side.
